```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Load Data (Optimized Caching)
# -----------------------------
@st.cache_data
def load_data():
    return pd.read_csv("Students Performance Dataset.csv")

df = load_data()

st.title("📊 Students Performance Dashboard")

# -----------------------------
# KPI Section
# -----------------------------
col1, col2, col3 = st.columns(3)

col1.metric("Total Students", len(df))
col2.metric("Average Total Score", round(df["Total_Score"].mean(), 2))
col3.metric("Highest Score", round(df["Total_Score"].max(), 2))

st.divider()

# -----------------------------
# Department Analysis
# -----------------------------
st.subheader("Average Total Score by Department")

dept_avg = df.groupby("Department")["Total_Score"].mean()

fig1 = plt.figure()
dept_avg.plot(kind="bar")
plt.xlabel("Department")
plt.ylabel("Average Total Score")
plt.title("Department Performance")
st.pyplot(fig1)

# -----------------------------
# Study Hours vs Total Score
# -----------------------------
st.subheader("Study Hours vs Total Score")

fig2 = plt.figure()
plt.scatter(df["Study_Hours_per_Week"], df["Total_Score"])
plt.xlabel("Study Hours per Week")
plt.ylabel("Total Score")
plt.title("Study Hours Impact")
st.pyplot(fig2)

# -----------------------------
# Attendance vs Total Score
# -----------------------------
st.subheader("Attendance vs Total Score")

fig3 = plt.figure()
plt.scatter(df["Attendance (%)"], df["Total_Score"])
plt.xlabel("Attendance (%)")
plt.ylabel("Total Score")
plt.title("Attendance Impact")
st.pyplot(fig3)

# -----------------------------
# Grade Distribution
# -----------------------------
st.subheader("Grade Distribution")

grade_counts = df["Grade"].value_counts()

fig4 = plt.figure()
grade_counts.plot(kind="bar")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.title("Grade Breakdown")
st.pyplot(fig4)
```
